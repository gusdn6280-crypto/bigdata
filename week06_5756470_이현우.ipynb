## 과제 내용
은닉층이 1개인 모델과 2개인 모델을 각각 작성하고 구조 차이를 설명합니다.

## 제출 방법
모든 셀을 실행해 출력이 보이는 상태로 저장한 뒤 `week06_학번_이름.ipynb` 로 제출합니다.


```python
import torch
import torch.nn as nn
import pandas as pd

torch.manual_seed(0)
```

## 문제 1. 은닉층 1개 모델 (25점)
`nn.Module` 을 상속해 아래 조건의 모델을 작성하세요.

- 입력 특성 5개, 은닉층 뉴런 16개, 출력 3개
- `__init__` 과 `forward` 를 모두 구현하고 각 줄에 주석
- `print(model)` 로 구조 출력


```python
# 답안
class SingleHiddenLayerModel(nn.Module):
    def __init__(self):
        # 부모 클래스(nn.Module)의 초기화 메서드를 호출
        super().__init__()
        # 입력 특성 5개를 받아 은닉층 뉴런 16개로 변환하는 선형 레이어를 정의
        self.fc1 = nn.Linear(5, 16)
        # 은닉층에 적용할 활성화 함수로 ReLU를 사용
        self.relu = nn.ReLU()
        # 은닉층의 16개 출력값을 받아서 최종 클래스 3개로 출력하는 선형 레이어를 정의
        self.fc2 = nn.Linear(16, 3)

    def forward(self, x):
        # 첫 번째 선형 레이어를 통과
        x = self.fc1(x)
        # 활성화 함수 ReLU를 적용
        x = self.relu(x)
        # 출력층 레이어를 통과시킨 최종 결과를 반환.
        x = self.fc2(x)
        return x

# 모델 객체 생성
model = SingleHiddenLayerModel()

# 모델 구조 출력
print(model)
```

## 문제 2. 은닉층 2개 모델 (25점)
문제 1과 같은 입력·출력에 은닉층을 2개로 늘린 모델을 작성하고 구조를 출력하세요.


```python
class TwoHiddenLayerModel(nn.Module):
    def __init__(self):
        # 부모 클래스(nn.Module)의 초기화 메서드를 호출한다
        super().__init__()
        # 입력 특성 5개를 받아 첫 번째 은닉층 뉴런 16개로 변환한다
        self.fc1 = nn.Linear(5, 16)
        # 첫 번째 은닉층의 16개 출력을 두 번째 은닉층 뉴런 8개로 줄여준다
        self.fc2 = nn.Linear(16, 8)
        # 두 번째 은닉층 8개 출력을 최종 클래스 3개로 출력한다
        self.fc3 = nn.Linear(8, 3)
        # 활성화 함수로 ReLU를 사용한다
        self.relu = nn.ReLU()

    def forward(self, x):
        # 첫 번째 은닉층 통과 후 ReLU 적용
        x = self.relu(self.fc1(x))
        # 두 번째 은닉층 통과 후 ReLU 적용
        x = self.relu(self.fc2(x))
        # 최종 출력층 통과
        x = self.fc3(x)
        return x

# 모델 객체 생성
model2 = TwoHiddenLayerModel()

# 모델 구조 출력
print(model2)
```

## 문제 3. 두 모델의 구조 비교표 (30점)
아래 항목으로 표를 만들어 출력하세요.

| 항목 | 은닉층 1개 | 은닉층 2개 |
| --- | --- | --- |
| Linear 층 수 | | |
| 전체 파라미터 수 | | |
| 활성화 함수 개수 | | |
| 출력 shape | | |


```python
# 파라미터 수를 계산하는 함수
def count_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

# 은닉층 1개 모델 파라미터 수: (5*16 + 16) + (16*3 + 3) = 96 + 51 = 147
params_m1 = count_parameters(model)
# 은닉층 2개 모델 파라미터 수: (5*16 + 16) + (16*8 + 8) + (8*3 + 3) = 96 + 136 + 27 = 259
params_m2 = count_parameters(model2)

# 출력용 표 작성
print(f"{'항목':<15} | {'은닉층 1개':<12} | {'은닉층 2개':<12}")
print("-" * 45)
print(f"{'Linear 층 수':<15} | {2:<12} | {3:<12}")
print(f"{'전체 파라미터 수':<15} | {params_m1:<12} | {params_m2:<12}")
print(f"{'활성화 함수 개수':<15} | {1:<12} | {2:<12}")
print(f"{'출력 shape':<15} | {'(batch_size, 3)':<12} | {'(batch_size, 3)':<12}")
```

## 문제 4. 구조 차이 설명 (20점)
아래 세 가지를 각각 두세 줄로 설명하세요.

1. 층을 늘리면 파라미터 수가 왜 늘어나는가
2. 활성화 함수가 없으면 층을 쌓아도 소용없는 이유
3. 층을 늘리는 것이 항상 좋은가 (근거를 함께)


**답:**

1. 층을 추가하면 이전 층의 뉴런들과 다음 층의 뉴련들이 서로 모두 연결되는 선형 연산이 늘어난다. 이 과장에서 각 연결마다 새로운 가중치 매트릭스와 뉴런별 편향 벡터가 추가되므로 층의 개수가 증가함에 따라 학습해야 할 전체 파라미터 수가 자연스럽게 증가한다.

2.활성화 함수 없이 선형층만 여러 개 쌓으면, 아무리 깊게 연결해도 복합 행렬 곱에 불과하여 결국 단 하나의 선형층으로 표현 가능해진다.

3.층을 늘리는 것이 항상 좋지는 않다. 층이 깊어지면 모델의 표현력은 좋아지지만, 훈련 데이터에 너무 과하게 적합되어 새로운 데이터에서 성능이 떨어지는 과적합 위험이 커지며 학습 중 기울기가 사라지는 문제가 발생할 수 있다. 연산량과 메모리 사용량이 크게 증가하므로 데이터의 크기와 문제의 난이도에 맞는 적절한 깊이를 찾는 것이 중요하다.


## 문제 5. 임의 입력으로 확인 (보너스)
