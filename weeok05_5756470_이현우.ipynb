````python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

torch.manual_seed(0)
````

## 문제 1. 사용자 정의 Dataset 작성 (40점)
아래 조건으로 Dataset 클래스를 작성하세요.

- 데이터 개수는 30개, 입력 특성은 4개
- `__init__` `__len__` `__getitem__` 세 메서드를 모두 구현
- 각 메서드가 하는 일을 주석으로 한 줄씩 설명


````python
# 답안
class CustomDataset(Dataset):
    def __init__(self):
        # 30개의 데이터 샘플과 4개의 입력 특성을 가진 데이터 X와 정답 y 생성
        self.x_data = torch.randn(30, 4)
        self.y_data = torch.randn(30, 1)

    def __len__(self):
        # 전체 데이터셋의 총 샘플 개수 반환
        return len(self.x_data)

    def __getitem__(self, idx):
        # 인덱스(idx)에 해당하는 데이터 한 개와 정답 라벨 한 쌍 반환
        return self.x_data[idx], self.y_data[idx]

# 데이터셋 객체 생성
dataset = CustomDataset()
````

## 문제 2. DataLoader로 batch 출력 (30점)
문제 1의 Dataset을 DataLoader에 넣고 아래를 출력하세요.

- batch 개수
- 각 batch의 입력 shape과 정답 shape
- 마지막 batch의 크기가 다른 이유 (batch size를 7로 두고 확인)


````python
# 답안
dataloader = DataLoader(dataset, batch_size=7, shuffle=False)

# 1. 전체 batch 개수 출력
print(f"전체 batch 개수: {len(dataloader)}")
print("-" * 40)

# 2. 각 batch의 입력 shape과 정답 shape 출력
for i, (batch_x, batch_y) in enumerate(dataloader):
    print(f"Batch {i+1} - 입력 shape: {batch_x.shape}, 정답 shape: {batch_y.shape}")

print("-" * 40)

# 3. 마지막 batch의 크기가 다른 이유 출력
print("마지막 batch의 크기가 다른 이유:")
print("전체 데이터(30개)를 batch_size(7)로 나눈 몫이 4이고 남은 데이터가 2개이기 때문입니다.")
print("앞의 4개 배치에는 7개씩 들어가고, 마지막 5번째 배치에는 남은 2개만 담기게 됩니다.")
````

## 문제 4. 학습 루프 작성 (20점)
문제 1의 Dataset으로 아래를 수행하세요.

1. `nn.Linear` 모델 하나를 만든다
2. 다섯 단계(예측 → 손실 → 초기화 → 역전파 → 수정)로 학습 루프를 작성한다
3. epoch마다 평균 손실을 출력해 손실이 내려가는지 확인한다


````python
# 답안
# 1. 모델, 손실함수, 옵티마이저 정의
model = nn.Linear(4, 1)
criterion = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# DataLoader 설정
dataloader = DataLoader(dataset, batch_size=4, shuffle=True)

epochs = 100

# 2. 5단계 학습 루프 작성
for epoch in range(1, epochs + 1):
    total_loss = 0.0

    for batch_x, batch_y in dataloader:
        # [1단계] 예측 (Forward)
        prediction = model(batch_x)

        # [2단계] 손실 계산 (Loss)
        loss = criterion(prediction, batch_y)

        # [3단계] 기울기 초기화 (Zero Grad)
        optimizer.zero_grad()

        # [4단계] 역전파 (Backward)
        loss.backward()

        # [5단계] 가중치 수정 (Step)
        optimizer.step()

        total_loss += loss.item()

    # 3. epoch마다 평균 손실 출력
    avg_loss = total_loss / len(dataloader)
    print(f"Epoch {epoch:3d}/{epochs} - Loss: {avg_loss:.4f}")
````
